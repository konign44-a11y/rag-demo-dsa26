```python
from ollama import embeddings, chat
from numpy import array, dot
from numpy.linalg import norm


EMBEDDING_MODEL = "qwen3-embedding:latest"
LANGUAGE_MODEL = "qwen3.5:9b"


def cosine_similarity(a, b):
    return round(dot(a, b) / (norm(a) * norm(b)), 2)



prompt = "Schreibe eine fiktive Geschichte aus einer anderen Galaxie über Herbst und Blätter"

text_2 = "Im Herbst fallen immer wieder Blätter zu Boden, aber dabei ändern sie ihre Farbe zu Blau"
text_3 = "Beim Rennen quietschten die Reifen der Autos"
```

```python
response_1 = embeddings(model=EMBEDDING_MODEL, prompt=prompt)
response_2 = embeddings(model=EMBEDDING_MODEL, prompt=text_2)
response_3 = embeddings(model=EMBEDDING_MODEL, prompt=text_3)

emb_p = response_1.embedding
emb_2 = response_2.embedding
emb_3 = response_3.embedding
```

```python
print(
    "Wie ähnlich sind Prompt und Text2?",
    cosine_similarity(emb_p, emb_2)
 ) # Similarity between text_1 and text_2

print(
    "Wie ähnlich sind Prompt und Text3?",
    cosine_similarity(emb_p, emb_3)
 ) # Similarity between text_1 and text_3
```

```python
rag_prompt = f"""Du bist ein hilfreicher Assistent. Bitte beantworte die folgende Frage: {prompt}. Nutze dazu die folgenden Informationen: '{text_2}'."""
```
