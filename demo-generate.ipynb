```python
from ollama import chat 


MODEL = "llama3.1:8b"

prompt = "Schreibe ein Herbstgedicht"

messages = [
    {"role": "user", "content": "Schreibe ein Herbstgedicht mit 5 Absätzen"},
    ]
```

```python
# Antwort vom Modell abrufen
response = chat(model=MODEL, messages=messages)
```

```python
print(response.message.content)
```

## Kontext erweitern

```python
# Response des Systems in den Messages ergänzen, damit der Kontext für die nächste Anfrage erhalten bleibt
messages.append(
    {"role": "assistant", "content": response.message.content}
)
```

```python
# ev. nächste User-Message ergänzen
messages.append(
    {"role": "user", "content": "Kürze das Gedicht auf 3 Absätze"}
)
```

```python
# Ausgabe der Antwort
chat(
    model=MODEL,
    messages=messages       
)
```
